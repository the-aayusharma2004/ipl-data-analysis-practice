# IPL Data Analysis (2008-2017)

This notebook explores the IPL **star schema** (gold layer) produced by the pipeline.
It reuses the query functions in `src/analysis.py` so the logic stays in one place —
the notebook only loads data and renders charts.

**Prerequisite:** run the pipeline first so the gold Parquet exists:
```
cd src && python pipeline.py
```

````python
import sys
from pathlib import Path

# Make the src/ modules importable from the notebooks/ folder.
sys.path.insert(0, str(Path.cwd().parent / "src"))

import matplotlib.pyplot as plt
import seaborn as sns

import analysis as A

sns.set_theme(style="whitegrid")

# Load all star-schema tables once and reuse across cells.
tables = A._load()
{name: df.shape for name, df in tables.items()}
````

## 1. Team performance

````python
wins = A.wins_per_team(tables)
ax = sns.barplot(data=wins.iloc[::-1], x="wins", y="team", hue="team", legend=False, palette="viridis")
ax.set_title("Matches won by team (2008-2017)"); ax.set_xlabel("Wins"); ax.set_ylabel("")
plt.show()
wins
````

````python
toss = A.toss_win_correlation(tables)
ax = sns.lineplot(data=toss, x="season", y="toss_winner_won_pct", marker="o")
ax.set_title("Toss winner also won the match (%)"); ax.set_ylabel("% of matches")
plt.show()
toss
````

## 2. Batting

````python
runs = A.top_run_scorers(tables=tables)
ax = sns.barplot(data=runs.iloc[::-1], x="runs", y="Player_Name", hue="Player_Name", legend=False, palette="viridis")
ax.set_title("Top run scorers"); ax.set_xlabel("Runs"); ax.set_ylabel("")
plt.show()
runs
````

````python
sr = A.top_strike_rates(tables=tables)
display(sr)
bnd = A.boundary_counts(tables=tables)
bnd
````

## 3. Bowling

````python
wk = A.top_wicket_takers(tables=tables)
ax = sns.barplot(data=wk.iloc[::-1], x="wickets", y="Player_Name", hue="Player_Name", legend=False, palette="rocket")
ax.set_title("Top wicket takers"); ax.set_xlabel("Wickets"); ax.set_ylabel("")
plt.show()
wk
````

````python
display(A.best_economy_rates(tables=tables))

dm = A.dismissal_type_mix(tables)
plt.pie(dm["count"], labels=dm["dismissal_type"], autopct="%1.1f%%", startangle=90)
plt.title("Dismissal type breakdown"); plt.show()
dm
````

## 4. Season / venue trends

````python
rps = A.runs_per_season(tables)
ax = sns.lineplot(data=rps, x="season", y="total_runs", marker="o")
ax.set_title("Total runs per season"); ax.set_ylabel("Runs")
plt.show()
rps
````

````python
display(A.top_scoring_venues(tables=tables))

pp = A.powerplay_run_share(tables)
ax = sns.lineplot(data=pp, x="season", y="powerplay_pct", marker="o")
ax.set_title("Powerplay run share per season"); ax.set_ylabel("% of runs in overs 1-6")
plt.show()
pp
````
